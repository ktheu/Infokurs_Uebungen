| Ausdruck | Bedeutung |
|:--|:--|
| `[0] * 5` | Liste aus fünf Nullen |
| `a.count(5)` | wie oft die Zahl 5 in der Liste `a` vorkommt |
| `s.count('e')` | wie oft das Zeichen `'e'` im String `s` vorkommt |

| `a.index(5)` | Index der ersten 5 in der Liste `a` |
| `a.remove(5)` | entfernt die erste 5 aus der Liste `a` |
| `a.pop(2)` | entfernt das Element mit Index 2 aus der Liste `a` und gibt es zurück |
| `s.split(', ')` | teilt den String `s` an jedem `', '` in eine Liste von Strings auf |
| `' '.join(a)` | fügt die Strings der Liste `a` zu einem String zusammen, mit Leerzeichen dazwischen |

`count`, `index`, `remove`, `pop`, `split` und `join` sind **Methoden**. Wie bei `append()` schreiben wir sie mit einem Punkt
hinter die Liste bzw. den String.

### Listen vervielfachen

Mit dem Operator `*` erzeugen wir eine Liste, in der dieselben Elemente mehrmals hintereinander stehen.
So bekommen wir schnell eine Liste mit einer bestimmten Länge, zum Beispiel für ein Spielfeld.

````python
a = [0] * 5
print(a)
````

````python
zeile = ['.'] * 8
print(zeile)
print(len(zeile))
````

````python
b = [1, 2] * 3
print(b)
````

Danach können wir die Elemente wie gewohnt über den Index ändern.

````python
zeile = ['.'] * 8
zeile[0] = 'R'
zeile[-1] = 'Z'
print(zeile)
````

##### Eine Liste mit Zählern

Eine häufige Anwendung ist eine Liste mit **Zählern**. Wir wollen wissen, wie oft jede Augenzahl gewürfelt wurde.
Dazu legen wir eine Liste mit sieben Nullen an. `zaehler[k]` zählt, wie oft die Augenzahl `k` vorkommt.
Den Index 0 brauchen wir nicht, so passt der Index genau zur Augenzahl.

````python
wuerfe = [3, 6, 1, 3, 3, 6]
zaehler = [0] * 7
for w in wuerfe:
    zaehler[w] += 1
print(zaehler)
````

Die 1 wurde einmal gewürfelt, die 3 dreimal und die 6 zweimal.

### Zählen mit count

Die Methode `count()` liefert, wie oft ein Wert in einer Liste vorkommt. Kommt er gar nicht vor, ist das Ergebnis 0.

````python
a = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]
print(a.count(5))
print(a.count(1))
print(a.count(7))
````

Bei einem String zählt `count()`, wie oft ein Zeichen vorkommt. Wir können auch nach längeren Teilen suchen.
Groß- und Kleinschreibung werden dabei unterschieden. Wollen wir beides zählen, wandeln wir den String vorher mit `lower()` in Kleinbuchstaben um.

````python
s = 'Erdbeermarmelade'
print(s.count('e'))
print(s.count('E'))
print(s.lower().count('e'))
print(s.count('ar'))
````

### Suchen mit in und index

Mit `in` prüfen wir, ob ein Wert in einer Liste vorkommt. Das Ergebnis ist `True` oder `False`.
Mit `not in` prüfen wir das Gegenteil. Das geht auch bei Strings.

````python
namen = ['Ben', 'Lena', 'Tom', 'Lena']
print('Tom' in namen)
print('Paul' in namen)
print('Paul' not in namen)
print('ss' in 'Mississippi')
````

Die Methode `index()` liefert den **Index**, an dem ein Wert zum ersten Mal in der Liste steht.

````python
namen = ['Ben', 'Lena', 'Tom', 'Lena']
print(namen.index('Lena'))
print(namen.index('Tom'))
````

Kommt der Wert nicht vor, bricht das Programm mit einem **ValueError** ab. Probiere es aus:

````python
namen = ['Ben', 'Lena', 'Tom', 'Lena']
print(namen.index('Paul'))
````

Deshalb prüfen wir vorher mit `in`, ob der Wert vorkommt:

````python
namen = ['Ben', 'Lena', 'Tom', 'Lena']
name = 'Paul'
if name in namen:
    print(name, 'steht an der Stelle', namen.index(name))
else:
    print(name, 'ist nicht in der Liste')
````

### Entfernen mit remove und pop

Die Methode `remove()` entfernt einen **Wert** aus der Liste. Kommt er mehrfach vor, wird nur das **erste** Vorkommen entfernt.
Kommt er gar nicht vor, gibt es wie bei `index()` einen ValueError.

````python
a = [5, 3, 8, 3, 1]
a.remove(3)
print(a)
````

Die Methode `pop()` entfernt das Element an einem **Index** und gibt es zurück. So können wir mit dem entfernten Element weiterarbeiten.

````python
a = [5, 3, 8, 3, 1]
x = a.pop(2)
print(x)
print(a)
````

Ohne Index entfernt `pop()` das **letzte** Element. Das ist praktisch bei einem Stapel, von dem man immer das oberste Element nimmt.

````python
stapel = ['Herz 7', 'Pik Dame', 'Karo Ass']
oben = stapel.pop()
print(oben)
print(stapel)
````

| Methode | entfernt | gibt zurück |
|:--|:--|:--|
| `a.remove(x)` | das erste Element mit dem **Wert** `x` | nichts |
| `a.pop(i)` | das Element mit dem **Index** `i` | das entfernte Element |
| `a.pop()` | das letzte Element | das entfernte Element |

Um **alle** Vorkommen eines Wertes zu entfernen, benutzen wir eine while-Schleife:

````python
a = [0, 4, 0, 0, 7, 0]
while 0 in a:
    a.remove(0)
print(a)
````

**Vorsicht:** Entferne keine Elemente aus einer Liste, während eine for-Schleife über genau diese Liste läuft.
Dabei werden Elemente übersprungen. Nimm dafür eine while-Schleife wie oben.

### Strings aufteilen mit split

Die Methode `split()` teilt einen String an einem **Trennzeichen** auf und liefert eine Liste von Strings.
Das Trennzeichen selbst kommt in der Liste nicht mehr vor.

````python
s = 'Anna, Ben, Carla'
a = s.split(', ')
print(a)
print(len(a))
````

````python
datum = '24.12.2026'
print(datum.split('.'))
````

Ohne Angabe eines Trennzeichens teilt `split()` an den Leerzeichen auf. Mehrere Leerzeichen hintereinander zählen dabei wie eines.

````python
s = 'Der   Hund  bellt'
print(s.split())
````

Oft stehen mehrere Zahlen in einer Zeile, zum Beispiel in einer Datei oder bei einer Eingabe mit `input()`.
Nach dem Aufteilen sind die Elemente noch **Strings**. Zum Rechnen wandeln wir sie mit `int()` in Zahlen um.

````python
zeile = '3 8 12 5'
teile = zeile.split()
print(teile)
zahlen = []
for t in teile:
    zahlen.append(int(t))
print(zahlen)
````

### Strings zusammenfügen mit join

Die Methode `join()` ist das Gegenstück zu `split()`: Sie fügt die Strings einer Liste zu **einem** String zusammen.
Vor dem Punkt steht der String, der zwischen die Elemente gesetzt wird.

````python
woerter = ['Das', 'ist', 'ein', 'Satz']
print(' '.join(woerter))
print('-'.join(woerter))
print(''.join(woerter))
````

So können wir zum Beispiel eine Zeile eines Spielfelds übersichtlich ausgeben:

````python
zeile = ['.'] * 8
zeile[2] = '#'
zeile[5] = '#'
print(''.join(zeile))
````

Alle Elemente der Liste müssen **Strings** sein, sonst gibt es einen TypeError. Zahlen wandeln wir vorher mit `str()` um.

````python
zahlen = [3, 8, 12]
teile = []
for z in zahlen:
    teile.append(str(z))
print(', '.join(teile))
````

`split()` und `join()` lassen sich gut kombinieren. Hier werden überflüssige Leerzeichen entfernt:

````python
s = 'zu   viele    Leerzeichen'
print(' '.join(s.split()))
````

```python
def ist_anagramm(w1, w2):
    w1 = w1.lower()
    w2 = w2.lower()
    if len(w1) != len(w2):
        return False
    for c in w1:
        if w1.count(c) != w2.count(c):
            return False
    return True
```

```python
kinder = ['Ali', 'Bea', 'Cem', 'Dana', 'Eva', 'Finn']
pos = 0
while len(kinder) > 1:
    pos = (pos + 2) % len(kinder)
    print(kinder.pop(pos), 'ist raus')
print(kinder[0], 'hat gewonnen')
```
