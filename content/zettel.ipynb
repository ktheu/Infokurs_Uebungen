### Zettel

Hier kannst du Python-Code ausprobieren.

**Code-Zelle erstellen**
- Klicke in der Werkzeugleiste oben auf **+**. Unter der aktuellen Zelle erscheint eine neue Code-Zelle.
- Oder: Mit **Esc** die Zelle verlassen (blauer Rand), dann **B** für eine neue Zelle darunter, **A** für eine neue Zelle darüber.

**Code-Zelle ausführen**
- **Shift + Enter**: Zelle ausführen und zur nächsten Zelle springen (gibt es keine, wird eine neue angelegt).
- **Strg + Enter**: Zelle ausführen und in der Zelle bleiben.
- Oder in der Werkzeugleiste auf **▶** klicken.

Die Ausgabe erscheint direkt unter der Zelle. Beim ersten Ausführen wird Python in den Browser geladen, das dauert ein paar Sekunden.
