#### Zur Erinnerung: Werkzeuge

| Was? | Python |
|:--|:--|
| Rest von $a$ bei der Division durch $n$ | `a % n` |
| $a^x$ in $\mathbb{Z}_n$ (Powermod-Verfahren) | `pow(a, x, n)` |
| Kehrwert $\frac{1}{k}$ in $\mathbb{Z}_n$ | `pow(k, -1, n)` (Fehlermeldung, wenn es keinen gibt) |
| ggT | `math.gcd(a, b)` |
| erweiterter euklidischer Algorithmus: $a \cdot x + b \cdot y = \text{ggT}(a, b)$ | `x, y, g = sympy.gcdex(a, b)` |
| Ist $p$ eine Primzahl? | `sympy.isprime(p)` |
| zufällige Primzahl zwischen $a$ und $b$ | `sympy.randprime(a, b)` |
| kleinste Primzahl größer als $n$ | `sympy.nextprime(n)` |
| Ist $k$ eine Primitivwurzel in $\mathbb{Z}_p$? | `sympy.is_primitive_root(k, p)` |
| kleinste Primitivwurzel in $\mathbb{Z}_p$ | `sympy.primitive_root(p)` |
| diskreter Logarithmus: $x$ mit $g^x = y$ in $\mathbb{Z}_p$ | `sympy.discrete_log(p, y, g)` |
| Primfaktorzerlegung von $m = p \cdot q$ | `p, q = sympy.factorint(m)` |
| SHA-256-Hashwert | `sha256(text.encode('utf-8')).hexdigest()` |

`sympy` ist eine Bibliothek für Mathematik. Führe die Zelle aus. Beim ersten Mal dauert das im Browser ein paar Sekunden.

#### Zur Erinnerung: Diffie-Hellman

#### Zur Erinnerung: RSA

#### Zur Erinnerung: Texte als Zahlen

RSA verschlüsselt Zahlen. Die beiden Funktionen wandeln einen Text in eine einzige große Zahl um und zurück. Dabei werden die UTF-8-Bytes des Textes einfach hintereinandergeschrieben. Führe die Zelle aus.

````python
def text_zu_zahl(text):
    return int.from_bytes(text.encode('utf-8'), 'big')

def zahl_zu_text(zahl):
    return zahl.to_bytes((zahl.bit_length() + 7) // 8, 'big').decode('utf-8')

n = text_zu_zahl('Hallo')
print(n)
print(zahl_zu_text(n))
````

#### Zur Erinnerung: Digitale Signatur

- Signieren mit dem privaten Schlüssel: $sig = hash(N)^d$ in $\mathbb{Z}_m$
- Prüfen mit dem öffentlichen Schlüssel: $sig^e$ in $\mathbb{Z}_m$ muss gleich $hash(N)$ sein

Unser RSA-Modul ist kleiner als der SHA-256-Hashwert. Deshalb nehmen wir vom Hashwert nur den Rest modulo $m$. `int(h, 16)` wandelt die Hexadezimalzahl `h` in eine ganze Zahl um. Führe die Zelle aus.

````python
def hashwert(nachricht, m):
    h = sha256(nachricht.encode('utf-8')).hexdigest()
    return int(h, 16) % m
````
