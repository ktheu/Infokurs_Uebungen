#### Zur Erinnerung: Digitale Signatur

- Signieren mit dem privaten Schlüssel: $sig = hash(N)^d$ in $\mathbb{Z}_m$
- Prüfen mit dem öffentlichen Schlüssel: $sig^e$ in $\mathbb{Z}_m$ muss gleich $hash(N)$ sein

Unsere RSA-Schlüssel sind sehr klein. Deshalb nehmen wir vom SHA-256-Hashwert nur den Rest modulo $m$. `int(h, 16)` wandelt die Hexadezimalzahl `h` in eine ganze Zahl um. Führe die Zelle aus.

````python
from hashlib import sha256

def hashwert(nachricht, m):
    h = sha256(nachricht.encode('utf-8')).hexdigest()
    return int(h, 16) % m
````
