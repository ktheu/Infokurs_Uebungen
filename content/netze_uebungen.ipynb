#### Erklärung: Adressen zerlegen mit split
Die String-Methode `split` zerlegt einen Text in eine **Liste** von Teilen. Ohne Angabe wird an den Leerzeichen getrennt. Man kann aber auch angeben, an welchem Zeichen getrennt werden soll, z. B. am Punkt einer IP-Adresse:

```python
'192.168.178.81'.split('.')     # ['192', '168', '178', '81']
```

Die Teile sind Texte. Mit `int` werden daraus Zahlen.

````python
ip = '192.168.178.81'
teile = ip.split('.')
print(teile)
print(int(teile[0]) + 1)
````

```python
def ist_gueltig(ip):
    '''
    ip: eine IP-Adresse als Text, z. B. '10.0.0.1'
    returns: True, wenn ip aus 4 Zahlen von 0 bis 255 besteht, sonst False

    >>> ist_gueltig('10.0.0.1')
    True

    >>> ist_gueltig('10.0.0.256')
    False
    '''
    teile = ip.split('.')
    if len(teile) != 4:
        return False
    for teil in teile:
        if int(teil) < 0 or int(teil) > 255:
            return False
    return True

print(ist_gueltig('10.0.1'))
```

#### Zur Erinnerung: Subnetzmaske und Suffix
Die Subnetzmaske besteht binär aus Einsen, gefolgt von Nullen. Der **Suffix** gibt die Anzahl der Einsen an:

```
255.255.255.0  = 11111111.11111111.11111111.00000000  ->  /24
```

Bei einem Suffix n bleiben 32 - n Bit für den Geräteteil. Davon sind die Adresse mit lauter Nullen (Netzadresse) und die mit lauter Einsen (Broadcast-Adresse) nicht für Geräte nutzbar.

```python
def suffix(maske):
    '''
    maske: eine Subnetzmaske als Text
    returns: die Anzahl der Einsen in der Binärdarstellung

    >>> suffix('255.255.255.0')
    24
    '''
    anzahl = 0
    for teil in maske.split('.'):
        anzahl += format(int(teil), '08b').count('1')
    return anzahl

print(suffix('255.255.255.0'))
print(suffix('255.255.248.0'))
print(suffix('255.255.255.224'))
```

```python
def gleiches_netz_24(ip1, ip2):
    '''
    ip1, ip2: zwei IP-Adressen als Text
    returns: True, wenn beide bei /24 im selben Netz liegen, sonst False

    >>> gleiches_netz_24('10.4.7.200', '10.4.7.3')
    True
    '''
    t1 = ip1.split('.')
    t2 = ip2.split('.')
    return t1[0] == t2[0] and t1[1] == t2[1] and t1[2] == t2[2]

print(gleiches_netz_24('10.4.7.200', '10.4.7.3'))
```
