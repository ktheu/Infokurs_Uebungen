#### Erklärung: Hex-Codes in Python
Eine Zahl von 0 bis 255 lässt sich in einem f-String mit `:02X` als zweistellige Hexadezimalzahl schreiben:

| Teil | Bedeutung |
|:--|:--|
| `X` | hexadezimal mit Großbuchstaben |
| `2` | mindestens 2 Stellen |
| `0` | links mit 0 auffüllen |

In die andere Richtung hilft `int(text, 16)`. Mit Slicing holst du dir die passenden zwei Zeichen aus dem Hex-Code: Bei `'#FF8000'` ist `[1:3]` der Rotanteil `'FF'`.

````python
r = 10
print(f'{r:02X}')
print(f'#{255:02X}{128:02X}{0:02X}')

farbe = '#FF8000'
print(farbe[1:3], int(farbe[1:3], 16))
````

#### Aufgabe A14
```
Bei der Lauflängenkodierung wird jede Lauflänge mit derselben Anzahl Bit
gespeichert. Schreibe eine Funktion kompressionsfaktor(laengen, bit), die
den Kompressionsfaktor zurückgibt:

    ursprüngliche Datenmenge = Summe der Lauflängen (1 Bit pro Pixel)
    komprimierte Datenmenge  = Anzahl der Lauflängen * bit
    Kompressionsfaktor       = komprimiert / ursprünglich
```
**Erwartete Ausgabe** (der Beispielaufrufe):
```

1.6
```

````python
def kompressionsfaktor(laengen, bit):
    '''
    laengen: Liste der Lauflängen
    bit: Anzahl Bit pro Lauflänge
    returns: der Kompressionsfaktor

    >>> kompressionsfaktor([4, 6, 7, 6, 1], 3)
    0.625
    '''
    # your code

print(kompressionsfaktor([4, 6, 7, 6, 1], 3))
print(kompressionsfaktor([1, 1, 1, 1, 1, 1, 2, 2], 2))
````

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def kompressionsfaktor(laengen, bit):
    '''
    laengen: Liste der Lauflängen
    bit: Anzahl Bit pro Lauflänge
    returns: der Kompressionsfaktor

    >>> kompressionsfaktor([4, 6, 7, 6, 1], 3)
    0.625
    '''
    urspruenglich = 0
    for laenge in laengen:
        urspruenglich += laenge
    komprimiert = len(laengen) * bit
    return komprimiert / urspruenglich

print(kompressionsfaktor([4, 6, 7, 6, 1], 3))
print(kompressionsfaktor([1, 1, 1, 1, 1, 1, 2, 2], 2))
```

#### Aufgabe A15

```
Schreibe eine Funktion lauflaengen(pixel), die zu einem String aus 0 und 1 die
Liste der Lauflängen zurückgibt (Start-Farbe Weiß = 1).

Tipp: Merke dir die aktuelle Farbe und zähle mit. Ist das nächste Pixel gleich,
erhöhst du den Zähler. Sonst hängst du den Zähler an die Liste an, wechselst
die Farbe und beginnst wieder bei 1. Am Ende muss der letzte Zähler noch an
die Liste.
Beginnt das Bild mit Schwarz, ist die erste Lauflänge 0. Das klappt
automatisch, wenn du mit Farbe '1' und Zähler 0 beginnst.
```
**Erwartete Ausgabe** (der Beispielaufrufe):
```
[3, 4, 6, 2]
[0, 2, 3, 2]
```

````python
def lauflaengen(pixel):
    '''
    pixel: String aus 0 und 1
    returns: Liste der Lauflängen, Start-Farbe Weiß (1)

    >>> lauflaengen('111000011111100')
    [3, 4, 6, 2]
    '''
    # your code

print(lauflaengen('111000011111100'))
print(lauflaengen('0011100'))
````

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def lauflaengen(pixel):
    '''
    pixel: String aus 0 und 1
    returns: Liste der Lauflängen, Start-Farbe Weiß (1)

    >>> lauflaengen('111000011111100')
    [3, 4, 6, 2]
    '''
    ergebnis = []
    farbe = '1'
    anzahl = 0
    for p in pixel:
        if p == farbe:
            anzahl += 1
        else:
            ergebnis.append(anzahl)
            farbe = p
            anzahl = 1
    ergebnis.append(anzahl)
    return ergebnis

print(lauflaengen('111000011111100'))
print(lauflaengen('0011100'))
```
