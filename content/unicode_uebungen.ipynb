#### Erklärung: ord und chr
Jedes Zeichen hat eine Nummer, den **Code Point**. Für die Zeichen aus der ASCII-Tabelle ist das genau der ASCII-Code.

| Funktion | Bedeutung | Beispiel | Ergebnis |
|:--|:--|:--|:--|
| `ord(zeichen)` | Zeichen -> Code Point | `ord('A')` | `65` |
| `chr(zahl)` | Code Point -> Zeichen | `chr(65)` | `'A'` |

`ord` bekommt genau **ein** Zeichen. `ord('AB')` gibt einen Fehler.

##### ASCII-Tabelle (druckbare Zeichen)
Code = Zahl am Zeilenanfang + Zahl über der Spalte. Beispiel: `A` steht in Zeile **64** und Spalte **+1**, hat also den Code 65. `SP` ist das Leerzeichen (Code 32).

#### Erklärung: Bitfolgen in Zeichen umwandeln
Ein Byte als Text, z. B. `'01000001'`, wandelst du mit `int(text, 2)` in eine Zahl um. Mit `chr` wird daraus das Zeichen.

Stehen mehrere Bytes durch Leerzeichen getrennt in einem Text, hilft die String-Methode `split`. Sie zerlegt einen Text an den Leerzeichen in eine **Liste** von Teilen:

```python
'01000001 01000010'.split()     # ['01000001', '01000010']
```

````python
bitfolge = '01000001 01000010'
teile = bitfolge.split()
print(teile)
print(chr(int(teile[0], 2)))
````

#### Erklärung: Zeichen außerhalb von ASCII
`ord` und `chr` funktionieren für **alle** Unicode-Zeichen, also auch für Umlaute, griechische Buchstaben oder Emojis. Code Points werden meist hexadezimal angegeben. In Python schreibst du sie mit `0x`:

```python
hex(ord('€'))      # '0x20ac'
chr(0x20AC)        # '€'
```

````python
print(ord('ö'), hex(ord('ö')))
print(chr(0x3C0))
print(chr(0x1F600))
````

#### Erklärung: UTF-8 in Python
Mit der String-Methode `encode('utf-8')` wird ein Text in seine UTF-8-Bytes umgewandelt. Mit `hex()` werden die Bytes hexadezimal angezeigt, mit `len` zählst du sie.

| Befehl | Ergebnis |
|:--|:--|
| `'€'.encode('utf-8').hex()` | `'e282ac'` |
| `len('€')` | `1` (Zeichen) |
| `len('€'.encode('utf-8'))` | `3` (Bytes) |

In die andere Richtung geht es mit `bytes.fromhex(...)` und `decode('utf-8')`:

```python
bytes.fromhex('e2 82 ac').decode('utf-8')     # '€'
```

````python
print('€'.encode('utf-8').hex())
print(len('€'), len('€'.encode('utf-8')))
print(bytes.fromhex('e2 82 ac').decode('utf-8'))
````

Schreibe eine Funktion anzahl_bytes(z), die zurückgibt, wie viele Bytes das
Zeichen z in UTF-8 braucht. Benutze dafür ord und die Bereiche aus dem Unterricht:

    Code Point 0 bis 7F          ->  1 Byte
    Code Point 80 bis 7FF        ->  2 Byte
    Code Point 800 bis FFFF      ->  3 Byte
    Code Point 10000 bis 10FFFF  ->  4 Byte
```
**Erwartete Ausgabe** (der Beispielaufrufe):

```

````python
def anzahl_bytes(z):
    '''
    z: ein Zeichen
    returns: die Anzahl der Bytes von z in UTF-8

    >>> anzahl_bytes('a')
    1

    >>> anzahl_bytes('€')
    3
    '''
    # your code

print(anzahl_bytes('a'))
print(anzahl_bytes('ß'))
print(anzahl_bytes('€'))
print(anzahl_bytes('🙂'))
````

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def anzahl_bytes(z):
    '''
    z: ein Zeichen
    returns: die Anzahl der Bytes von z in UTF-8

    >>> anzahl_bytes('a')
    1

    >>> anzahl_bytes('€')
    3
    '''
    c = ord(z)
    if c <= 0x7F:
        return 1
    elif c <= 0x7FF:
        return 2
    elif c <= 0xFFFF:
        return 3
    else:
        return 4

print(anzahl_bytes('a'))
print(anzahl_bytes('ß'))
print(anzahl_bytes('€'))
print(anzahl_bytes('🙂'))
```
