#### Aufgabe A12
```
Addiere mit einer for-Schleife die Zahlen ab 1 nacheinander zu einer Summe auf.
Überspringe dabei alle geraden Zahlen (mit continue).
Breche die Schleife ab (mit break), sobald die Summe größer als 100 ist.
Gib danach die zuletzt addierte Zahl und die Summe in einer Zeile aus.
```
**Erwartete Ausgabe:**
```
21 121
```

````python
# your code
````

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
summe = 0
for k in range(1, 100):
    if k % 2 == 0:
        continue
    summe += k
    if summe > 100:
        break
print(k, summe)
```

</details>

#### Aufgabe A16
```
Lies mit einer Schleife while True so lange ganze Zahlen ein,
bis eine negative Zahl eingegeben wird. Beende die Schleife dann mit break.
Gib danach aus, wie viele Zahlen eingegeben wurden (ohne die negative Zahl)
und wie groß ihre Summe ist.
```
**Beispieldialog:**
```
Zahl: 5
Zahl: 3
Zahl: 10
Zahl: -1
Anzahl: 3, Summe: 18
```

````python
# your code
````

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
anzahl = 0
summe = 0
while True:
    zahl = int(input('Zahl: '))
    if zahl < 0:
        break
    anzahl += 1
    summe += zahl
print(f'Anzahl: {anzahl}, Summe: {summe}')
```
