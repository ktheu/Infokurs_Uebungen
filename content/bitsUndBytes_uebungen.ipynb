#### Zur Erinnerung: Umrechnen in Python

| Was | Befehl | Ergebnis |
|:--|:--|:--|
| Binär geschriebene Zahl | `0b1101` | `13` |
| Hexadezimal geschriebene Zahl | `0x2F` | `47` |
| Text im Zahlsystem mit Basis 2 / 16 -> Zahl | `int('1101', 2)`, `int('2F', 16)` | `13`, `47` |
| Zahl -> Binärtext | `bin(13)` | `'0b1101'` |
| Zahl -> Hextext | `hex(47)` | `'0x2f'` |
| Präfix weglassen | `bin(13)[2:]`, `hex(47)[2:]` | `'1101'`, `'2f'` |
| ganzzahlige Division / Modulo-Operator | `13 // 2`, `13 % 2` | `6`, `1` |

#### Neu: Binärzahlen mit fester Stellenzahl
`bin(5)[2:]` liefert `'101'`. Oft möchte man aber eine feste Anzahl von Stellen haben, zum Beispiel 8 Stellen für ein Byte. Das geht mit der Funktion `format`:

```python
format(5, '08b')     # '00000101'
```

Der zweite Wert `'08b'` ist ein Text und bedeutet:

| Teil | Bedeutung |
|:--|:--|
| `b` | als Binärzahl schreiben (ohne Präfix `0b`) |
| `8` | mindestens 8 Stellen |
| `0` | links mit Nullen auffüllen |

Mit `'016b'` bekommt man entsprechend 16 Stellen. Ist die Zahl länger, wird nichts abgeschnitten: `format(300, '08b')` ergibt `'100101100'` mit 9 Stellen.

````python
print(format(5, '08b'))
print(format(5, '016b'))
print(format(300, '08b'))
````

````python
def ist_binaer(s):
    '''
    s: ein Text
    returns: True, wenn s nur aus den Ziffern 0 und 1 besteht, sonst False

    >>> ist_binaer('101101')
    True
````

```
Schreibe eine Funktion bin2dez(s), die eine Binärzahl als Text bekommt und ihren
Wert als Zahl zurückgibt. Benutze dabei nicht int(s, 2).

Tipp: Gehe die Ziffern von links nach rechts durch. Der bisherige Wert wird mit 2
multipliziert und die neue Ziffer addiert:
    '1101':  0*2 + 1 = 1,  1*2 + 1 = 3,  3*2 + 0 = 6,  6*2 + 1 = 13
```
**Erwartete Ausgabe:**
```
13
32
255
```

````python
def bin2dez(s):
    '''
    s: eine Binärzahl als Text
    returns: der Wert von s als Zahl

    >>> bin2dez('1101')
    13
    '''
    # your code

print(bin2dez('1101'))
print(bin2dez('100000'))
print(bin2dez('11111111'))
````

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def bin2dez(s):
    '''
    s: eine Binärzahl als Text
    returns: der Wert von s als Zahl

    >>> bin2dez('1101')
    13
    '''
    wert = 0
    for ziffer in s:
        wert = wert * 2 + int(ziffer)
    return wert


print(bin2dez('1101'))
print(bin2dez('100000'))
print(bin2dez('11111111'))
```

<details>
<summary>Lösung anzeigen (erst selbst probieren!)</summary>

```python
def dez2bin(x):
    '''
    x: eine ganze Zahl größer als 0
    returns: die Binärdarstellung von x als Text

    >>> dez2bin(13)
    '1101'
    '''
    ergebnis = ''
    while x > 0:
        ergebnis = str(x % 2) + ergebnis
        x = x // 2
    return ergebnis


print(dez2bin(13))
print(dez2bin(64))
print(dez2bin(100))
```

#### Neu: Die String-Methode count
Wie Listen haben auch Strings Methoden. Mit `s.count(t)` zählt man, wie oft der Text `t` im String `s` vorkommt:

```python
'Banane'.count('a')      # 2
'101101'.count('1')      # 4
```

So kann man zum Beispiel zählen, wie viele Einsen die Binärdarstellung einer Zahl hat.

````python
print('Banane'.count('a'))
print('101101'.count('1'))
print(bin(100)[2:].count('1'))
````
