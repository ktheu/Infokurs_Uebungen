````python
wert = 5
print(type(wert))
wert = wert / 5
print(type(wert))
wert = str(wert)
print(type(wert))
print(wert * 3)
````

#### Aufgabe A12
```
In der Variable antwort steht als Text, ob jemand den Newsletter haben möchte:
'True' oder 'False'. Das Programm gibt aber in beiden Fällen 'Newsletter wird
verschickt' aus. Ändere die Bedingung so, dass das Programm richtig arbeitet.
Teste es mit beiden Werten von antwort.
```

**Erwartete Ausgabe:**
```
Kein Newsletter
```

````python
antwort = 'False'

if bool(antwort):
    print('Newsletter wird verschickt')
else:
    print('Kein Newsletter')
````
